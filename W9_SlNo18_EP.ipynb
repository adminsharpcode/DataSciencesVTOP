## Problem Statement
Create an array of integers with the values 2, 5, 9, 14, 20. Write a function that takes this array and returns the sum of its elements. Next, create an array of strings with the names of five different fruits. Write a function that concatenates all elements of this array, separated by commas. Finally, iterate over the array of integers and print each element to the console.

## Objectives
- To understand array memory allocation and contiguous indexed storage in the Scala programming language.
- To implement pure mathematical reduction functions for array aggregation including element summation.
- To manipulate collections of strings and perform delimiter-separated sequence formatting using mkString.
- To explore declarative traversal paradigms using the higher-order foreach combinator on JVM arrays.
- To evaluate the interplay between mutable array structures and functional immutability in Scala.


```scala
# --- Cell 1: Integer Array Summation and String Array Delimiter Concatenation ---
// Program 18: Arrays and Array Operations in Scala
// Student: Vedant Nimkar | Reg. No: 26MML0045

object Experiment18_Arrays {
  def main(args: Array[String]): Unit = {
    // 1. Array of integers with specified values
    val intArr: Array[Int] = Array(2, 5, 9, 14, 20)

    // Pure function to calculate array sum
    def sumArray(arr: Array[Int]): Int = {
      arr.sum
    }

    val totalSum = sumArray(intArr)
    println(s"Sum of integer array elements: $totalSum")

    // 2. Array of strings containing five fruit names
    val fruitArr: Array[String] = Array("Apple", "Banana", "Cherry", "Date", "Elderberry")

    // Pure function to concatenate strings separated by commas
    def concatenateFruits(arr: Array[String]): String = {
      arr.mkString(", ")
    }

    val fruitString = concatenateFruits(fruitArr)
    println(s"Concatenated fruits: $fruitString")
  }
}
```

## Conclusion
The Scala array operations experiment demonstrated the fundamental memory mechanics, syntactic elegance, and functional processing capabilities of Scala's Array data structure.

- Arrays in Scala map directly to native JVM primitive arrays, ensuring maximum memory efficiency and speed.
- Pure functional reduction methods like arr.sum provide safe and concise numerical calculations.
- The mkString combinator offers flexible sequence serialization with custom delimiters, prefixes, and suffixes.
- Higher-order traversal methods such as foreach encapsulate iteration logic without mutable index variables.
- Scala bridges imperative JVM data structures with functional programming elegance.

Mastering fundamental collection types like Arrays in Scala provides the essential conceptual basis for scalable big data engineering in modern distributed frameworks like Apache Spark.
