## Problem Statement
An e-commerce company maintains product sales records containing ProductID, SalesAmount, and Category. Create a DataFrame for 20 products and perform the following operations: select only the ProductID and SalesAmount columns, filter the products whose SalesAmount is greater than Rs. 50,000, add a new column Region with the default value "South India", sort the products by SalesAmount in descending order, and calculate the average SalesAmount of all the products.

## Objectives
- To create a Pandas DataFrame representing e-commerce product sales records.
- To select specific columns of business interest from a sales dataset.
- To filter high-value products using a conditional expression on SalesAmount.
- To add a new derived column with a constant default value.
- To rank products by sales performance using descending order sorting.
- To compute the average sales amount using Pandas aggregate functions.


```python
# --- Cell 1: Product Sales DataFrame of 20 Products ---
# Load Necessary Packages
import pandas as pd

# Step 1: Create the e-commerce product sales dataset
data = {
    'ProductID': ['P001', 'P002', 'P003', 'P004', 'P005',
                  'P006', 'P007', 'P008', 'P009', 'P010',
                  'P011', 'P012', 'P013', 'P014', 'P015',
                  'P016', 'P017', 'P018', 'P019', 'P020'],
    'SalesAmount': [72000, 45000, 88000, 32000, 95000,
                    51000, 28000, 67000, 41000, 79000,
                    56000, 34000, 91000, 47000, 62000,
                    38000, 84000, 53000, 29000, 76000],
    'Category': ['Electronics', 'Fashion', 'Home Appliances', 'Grocery', 'Footwear',
                 'Electronics', 'Fashion', 'Home Appliances', 'Grocery', 'Footwear',
                 'Electronics', 'Fashion', 'Home Appliances', 'Grocery', 'Footwear',
                 'Electronics', 'Fashion', 'Home Appliances', 'Grocery', 'Footwear']
}

# Step 2: Create the DataFrame
df = pd.DataFrame(data)

print("Product Sales DataFrame")
print(df)
print("\nShape of the DataFrame =", df.shape)
```

```python
# --- Cell 2: Selected ProductID and SalesAmount Columns ---
# Step 3: Select only ProductID and SalesAmount columns
product_sales = df[['ProductID', 'SalesAmount']]
print("\nProductID and SalesAmount")
print(product_sales)
```

```python
# --- Cell 3: Filtered High-Value Products (SalesAmount > Rs. 50,000) ---
# Step 4: Filter products with SalesAmount greater than 50000
high_sales = df[df['SalesAmount'] > 50000]
print("\nProducts with SalesAmount greater than Rs. 50,000")
print(high_sales)
print("\nNumber of high-value products =", len(high_sales))
```

```python
# --- Cell 4: Sales DataFrame with Region Column Added ---
# Step 5: Add a new column Region with default value
df['Region'] = 'South India'
print("\nDataFrame after adding the Region column")
print(df)
```

```python
# --- Cell 5: Products Sorted by SalesAmount (Descending Order) ---
# Step 6: Sort the products by SalesAmount in descending order
sorted_df = df.sort_values(by='SalesAmount', ascending=False)
print("\nProducts sorted by SalesAmount (Descending Order)")
print(sorted_df)
```

```python
# --- Cell 6: Average, Maximum and Minimum SalesAmount ---
# Step 7: Calculate the average SalesAmount
average_sales = df['SalesAmount'].mean()
print("\nAverage SalesAmount = Rs.", average_sales)
print("Maximum SalesAmount = Rs.", df['SalesAmount'].max())
print("Minimum SalesAmount = Rs.", df['SalesAmount'].min())
```
