## Problem Statement
Create a 2D array of shape (3, 4) with random integers between 0 and 10. Perform the following operations: calculate the mean and standard deviation of the entire array, slice the array to get the first two rows and the last two columns, reshape the array to shape (4, 3), and perform element-wise multiplication with another array of the same shape.

## Objectives
- To understand the creation of NumPy ndarray objects using random integers.
- To compute descriptive statistics such as mean and standard deviation of an array.
- To perform slicing operations in order to extract a sub-array.
- To reshape an array from one dimension configuration to another.
- To carry out element-wise arithmetic between two arrays of the same shape.
- To appreciate the efficiency of vectorized operations offered by NumPy.


```python
# --- Cell 1: Original 3 x 4 Array of Random Integers ---
# Load Necessary Packages
import numpy as np

# Step 1: Fix the random seed for reproducible results
np.random.seed(42)

# Step 2: Create a 2D array of shape (3, 4) with random integers 0 to 10
arr = np.random.randint(0, 11, size=(3, 4))

print("Original 2D Array of shape (3, 4)")
print(arr)
print("\nShape =", arr.shape)
print("Size  =", arr.size)
print("Dtype =", arr.dtype)
```

```python
# --- Cell 2: Mean and Standard Deviation of the Entire Array ---
# Step 3: Calculate the mean and standard deviation of the entire array
mean_value = arr.mean()
std_value = arr.std()

print("\nMean of all elements              =", mean_value)
print("Standard Deviation of all elements =", std_value)
```

```python
# --- Cell 3: Sliced Sub-array (First Two Rows, Last Two Columns) ---
# Step 4: Slice the first two rows and the last two columns
sliced = arr[:2, -2:]

print("\nSliced Array (first two rows, last two columns)")
print(sliced)
print("\nShape of the sliced array =", sliced.shape)
```

```python
# --- Cell 4: Array Reshaped from (3, 4) to (4, 3) ---
# Step 5: Reshape the array from (3, 4) to (4, 3)
reshaped = arr.reshape(4, 3)

print("\nReshaped Array of shape (4, 3)")
print(reshaped)
print("\nShape after reshaping =", reshaped.shape)
```

```python
# --- Cell 5: Second (4, 3) Array Used for Multiplication ---
# Step 6: Create another array of the same shape (4, 3)
arr2 = np.random.randint(1, 6, size=(4, 3))

print("\nSecond Array of shape (4, 3)")
print(arr2)
```

```python
# --- Cell 6: Element-wise Multiplication of the Two Arrays ---
# Step 7: Perform element-wise multiplication of the two (4, 3) arrays
product = reshaped * arr2

print("\nElement-wise Multiplication Result")
print(product)
print("\nVerification using np.multiply()")
print(np.multiply(reshaped, arr2))
```

## Conclusion
This experiment successfully demonstrated the creation and manipulation of multi-dimensional arrays using the NumPy library. A 3 x 4 array of random integers was generated with a fixed seed, and a sequence of statistical, slicing, reshaping, and arithmetic operations was applied to it. The results showed that:

- np.random.randint() together with np.random.seed() created a reproducible 3 x 4 array of random integers in the required range.
- The mean() and std() functions summarised the entire array into a measure of central tendency and a measure of dispersion respectively.
- Slicing with arr[:2, -2:] extracted the required 2 x 2 sub-array using positive and negative index ranges, returning an efficient view of the data.
- The reshape() function converted the (3, 4) array into a (4, 3) array in row-major order without copying or altering the underlying data.
- Element-wise multiplication with a second (4, 3) array produced the position-wise product, and the identical np.multiply() result confirmed the behaviour.

Overall, this exercise established a clear understanding of ndarray creation, descriptive statistics, indexing and slicing, shape manipulation, and vectorized arithmetic in NumPy. Because NumPy arrays are homogeneous and stored contiguously in memory, these operations execute far faster than equivalent pure Python list operations, which is why NumPy forms the numerical foundation of Pandas, scikit-learn, and virtually every scientific computing library in Python.
