## Problem Statement
Create an interactive scatter plot showing the relationship between petal length and petal width from the Iris dataset using Bokeh. Load the Iris dataset, configure a ColumnDataSource, map species to distinct colors and markers, attach an interactive HoverTool displaying sepal and petal measurements and species labels, configure interactive pan and wheel-zoom tools, and display the styled Bokeh figure.

## Objectives
- To master Bokeh's glyph-based plotting architecture and ColumnDataSource integration.
- To load and inspect multivariate botanical measurements from the Fisher Iris dataset.
- To build an interactive scatter plot using figure.scatter.
- To attach a configured HoverTool showing formatted feature values on mouseover.
- To apply categorical color mapping using Bokeh factor_cmap.
- To interpret species clustering and morphometric separation between Setosa, Versicolor, and Virginica.


```python
# --- Cell 1: Iris Botanical Dataset and Species Petal Morphometrics ---
# Load Necessary Packages
import pandas as pd
from sklearn.datasets import load_iris
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.transform import factor_cmap

# Step 1 & 2: Load Iris dataset into DataFrame
iris = load_iris()
df_iris = pd.DataFrame(iris.data, columns=["sepal_length", "sepal_width", "petal_length", "petal_width"])
species_names = ["setosa", "versicolor", "virginica"]
df_iris["species"] = [species_names[i] for i in iris.target]

print("=== Fisher's Iris Dataset (First 5 Rows) ===")
print(df_iris.head(5).to_string(index=False))

# Step 3: Compute group-level petal metrics
print("\n=== Petal Dimensions by Species (Means) ===")
species_stats = df_iris.groupby("species")[["petal_length", "petal_width"]].agg(["mean", "std", "min", "max"]).round(2)
print(species_stats)
```

```python
# --- Cell 2: Interactive Bokeh Scatter Plot with HoverTool and Factor Color Mapping ---
# Step 4 to 8: Interactive Bokeh Scatter Plot
source = ColumnDataSource(df_iris)

species_list = ["setosa", "versicolor", "virginica"]
palette = ["#1f77b4", "#ff7f0e", "#2ca02c"]
color_mapping = factor_cmap("species", palette=palette, factors=species_list)

hover = HoverTool(tooltips=[
    ("Species", "@species"),
    ("Petal Length", "@petal_length{0.0} cm"),
    ("Petal Width", "@petal_width{0.0} cm"),
    ("Sepal Length", "@sepal_length{0.0} cm"),
    ("Sepal Width", "@sepal_width{0.0} cm")
])

p = figure(
    title="Iris Petal Dimensions by Botanical Species (Bokeh Interactive)",
    x_axis_label="Petal Length (cm)",
    y_axis_label="Petal Width (cm)",
    width=680,
    height=440,
    tools=[hover, "pan", "wheel_zoom", "box_zoom", "reset"]
)

p.scatter(
    x="petal_length",
    y="petal_width",
    source=source,
    size=9,
    color=color_mapping,
    legend_field="species",
    alpha=0.85,
    line_color="black",
    line_width=0.5
)

p.title.text_font_size = "13pt"
p.title.align = "center"
p.legend.title = "Species"
p.legend.location = "top_left"
p.legend.click_policy = "hide"
p.background_fill_color = "#fbfbfb"
p.grid.grid_line_dash = [4, 4]
p.grid.grid_line_alpha = 0.5

show(p)
```
