## Problem Statement
Create an interactive bar chart showing the population of different cities. Define a dataset comprising major world cities with their respective population (in millions), continent, and country. Render an interactive Plotly bar chart with customized color mapping by continent, hover tooltips displaying full city details, clear axis labels, and formatted titles.

## Objectives
- To understand the architecture of interactive data visualization using Plotly Express.
- To construct structured population datasets with multi-level metadata (city, country, continent).
- To build an interactive bar chart using px.bar.
- To customize visual encodings, including discrete color maps, hover templates, and axis formatting.
- To sort and rank categories dynamically based on metric values.
- To analyze city demographic distribution and interpret continental urban concentration.


```python
# --- Cell 1: World Cities Demographic Dataset and Grouped Aggregations ---
# Load Necessary Packages
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Step 1 & 2: Construct demographic dataset
city_data = {
    "City": ["Tokyo", "Delhi", "Shanghai", "Dhaka", "São Paulo", "Mexico City",
             "Cairo", "Beijing", "Mumbai", "New York", "Paris", "London"],
    "Country": ["Japan", "India", "China", "Bangladesh", "Brazil", "Mexico",
                "Egypt", "China", "India", "United States", "France", "United Kingdom"],
    "Continent": ["Asia", "Asia", "Asia", "Asia", "South America", "North America",
                  "Africa", "Asia", "Asia", "North America", "Europe", "Europe"],
    "Population_Millions": [37.4, 32.9, 29.3, 23.2, 22.6, 22.3, 22.2, 21.5, 21.3, 18.8, 11.2, 9.6]
}

df_cities = pd.DataFrame(city_data)

# Step 3: Sort by population descending
df_cities = df_cities.sort_values("Population_Millions", ascending=False).reset_index(drop=True)

print("=== Global Metropolitan Populations Table ===")
print(df_cities.to_string(index=False))

# Step 4: Summary statistics & Continental aggregation
print("\n=== Population Descriptive Statistics (Millions) ===")
print(df_cities["Population_Millions"].describe().round(2))

print("\n=== Continental Population Aggregates ===")
cont_df = df_cities.groupby("Continent").agg(
    City_Count=("City", "count"),
    Total_Pop=("Population_Millions", "sum"),
    Avg_Pop=("Population_Millions", "mean")
).round(2).sort_values("Total_Pop", ascending=False)
print(cont_df)
```

```python
# --- Cell 2: Interactive Plotly Bar Chart Showing City Population by Continent ---
# Step 5 to 8: Create interactive Plotly bar chart
color_palette = {
    "Asia": "#2b6cb0",
    "South America": "#d69e2e",
    "North America": "#dd6b20",
    "Africa": "#38a169",
    "Europe": "#805ad5"
}

fig = px.bar(
    df_cities,
    x="City",
    y="Population_Millions",
    color="Continent",
    color_discrete_map=color_palette,
    title="<b>Global Metropolitan City Populations</b>",
    labels={"Population_Millions": "Population (Millions)", "City": "City Name"},
    text="Population_Millions",
    hover_data={"Country": True, "Continent": True, "Population_Millions": ":.1f"}
)

fig.update_traces(
    texttemplate="%{text:.1f}M",
    textposition="outside",
    marker_line_color="#1a202c",
    marker_line_width=1.0,
    hovertemplate="<b>%{x}</b> (%{customdata[0]})<br>Continent: %{customdata[1]}<br>Population: %{y:.1f} Million<extra></extra>"
)

fig.update_layout(
    title_font_size=18,
    title_x=0.5,
    xaxis_tickangle=-30,
    yaxis_range=[0, 42],
    yaxis_gridcolor="#e2e8f0",
    plot_bgcolor="white",
    paper_bgcolor="white",
    margin=dict(l=50, r=40, t=60, b=70),
    legend=dict(title="Continent", orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)

fig.show()
```
