## Problem Statement
Use Breeze library to calculate total revenue for each fruit type by multiplying price and quantity from a dataset. You will then create a bar chart to visualize these revenues using Plotly. Scala. This exercise introduces numerical computations with Breeze and dynamic visualizations with Plotly. Scala.

## Objectives
- To master numerical linear algebra and vectorization in Scala using the Breeze scientific computing library.
- To construct Breeze DenseVector data structures holding continuous numerical parameters.
- To execute vectorized element-wise multiplication (:*) between unit price and volume vectors.
- To compute aggregate summary metrics including total gross revenue and mean revenue per category.
- To construct interactive, dynamic bar chart visualizations using Plotly.Scala.


```scala
# --- Cell 1: Breeze Vectorized Element-Wise Multiplication and Revenue Computation ---
// Program 24: Numerical Computing with Breeze and Plotly.Scala
// Student: Vedant Nimkar | Reg. No: 26MML0045

import breeze.linalg._

object Experiment24_BreezePlotly {
  def main(args: Array[String]): Unit = {
    val fruits = Array("Apple", "Banana", "Orange", "Mango", "Grape")

    // Breeze DenseVectors for Price and Quantity
    val prices = DenseVector(120.0, 45.0, 80.0, 150.0, 95.0)
    val quantities = DenseVector(250.0, 600.0, 320.0, 180.0, 210.0)

    // Vectorized Element-wise Multiplication
    val revenues = prices :* quantities
    val totalGross = sum(revenues)
    val meanRevenue = mean(revenues)

    println("=== Breeze Numerical Computations: Total Revenue per Fruit ===") 
    println(s"Price Vector (INR):    $prices")
    println(s"Quantity Vector:       $quantities")
    println(s"Revenue Vector (INR):  $revenues\n")
    for (i <- 0 until fruits.length) {
      println(f"  ${fruits(i)}%-7s -> INR ${revenues(i)}%,10.2f")
    }
    println(f"\nTotal Gross Revenue: INR $totalGross%,10.2f")
    println(f"Mean Revenue/Fruit:  INR $meanRevenue%,10.2f")
  }
}
```

## Conclusion
The Breeze numerical computing and Plotly.Scala visualization experiment demonstrated the capabilities of high-performance linear algebra coupled with modern dynamic plotting in Scala.

- Breeze DenseVector leverages BLAS and LAPACK bindings for high-throughput vectorized math operations.

- Built-in aggregation functions (sum, mean, max) streamline mathematical analytics workflows.
- Plotly.Scala translates Scala data structures into responsive, interactive JavaScript D3/WebGL graphics.
- The combination of Breeze and Plotly.Scala delivers a robust numerical computing pipeline on the JVM.
