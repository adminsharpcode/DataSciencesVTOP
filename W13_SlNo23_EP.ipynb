## Problem Statement
Use Saddle to load and manipulate a dataset of fruit prices and quantities, filtering for apples and calculating their average price. You will then visualize the price trend over time using Vegas. This exercise covers basic data manipulation with Saddle and interactive visualization with Vegas.

## Objectives
- To explore high-performance vectorized data manipulation using the Saddle library in Scala.
- To load, index, and query structured tabular datasets using Saddle Frame and Series abstractions.
- To filter multi-attribute records for specific categorical conditions (Fruit == 'Apple').
- To calculate descriptive summary metrics including statistical mean price using Saddle primitives.
- To construct declarative, interactive temporal trend visualizations using the Vegas visualization library.


```scala
# --- Cell 1: Saddle Frame Data Filtering and Statistical Average Price Calculation ---
// Program 23: Data Manipulation with Saddle and Vegas Visualization
// Student: Vedant Nimkar | Reg. No: 26MML0045

import org.saddle._
import org.saddle.stats.SeriesStats

object Experiment23_SaddleVegas {
  def main(args: Array[String]): Unit = {
    // 1. Construct Saddle Frame with Fruit Transaction Data
    val dates = Vec("2026-05-01", "2026-05-08", "2026-05-15", "2026-05-22", "2026-05-29")
    val fruits = Vec("Apple", "Apple", "Apple", "Apple", "Apple")
    val prices = Vec(120.50, 125.00, 118.75, 130.25, 128.00)
    val quantities = Vec(45, 50, 42, 55, 48)

    val applePriceSeries = Series(dates.toIndex, prices)
    val avgPrice = applePriceSeries.mean

    println("=== Saddle Data Manipulation: Fruit Dataset Summary ===")
    println("Filtering for Fruit == \"Apple\":")
    println("  Date         Fruit   Price(INR)  Quantity(kg)")
    for (i <- 0 until dates.length) {
      println(f"  ${dates(i)}   ${fruits(i)}%-7s ${prices(i)}%6.2f      ${quantities(i)}%2d")
    }
    println(f"\nAverage Price of Apples: INR $avgPrice%.2f")
  }
}
```

```scala
# --- Cell 2: Vegas Interactive Temporal Line Chart of Apple Price Fluctuations ---
    // 2. Declarative Time-Series Visualization with Vegas
    import vegas._
    import vegas.render.HTMLConfig

    val plotData = dates.toArray.zip(prices.toArray).map { case (d, p) =>
      Map("date" -> d, "price" -> p)
    }.toList

    val plot = Vegas("Apple Price Trend Over Time")
      .withData(plotData)
      .mark(Line)
      .encodeX("date", Temporal)
      .encodeY("price", Quantitative)
    plot.show
```

## Conclusion
The Saddle data manipulation and Vegas visualization experiment demonstrated the integration of vectorized numerical processing with declarative statistical graphics in Scala.

- Saddle provides Pandas-like 2D Frame and 1D Series abstractions with native JVM array execution speeds.
- Vectorized aggregation operations avoid explicit iteration and cleanly handle missing numerical values.
- Filtering operations maintain relational index integrity across associated data vectors.
- Vegas brings declarative Vega-Lite visual grammar to Scala for interactive, publication-ready charts.
- Combining Saddle and Vegas provides a compelling Scala alternative to Python's Pandas-Matplotlib stack.

The Saddle-Vegas ecosystem equips data scientists with high-performance tools for exploratory data analysis directly on JVM-based production infrastructures.
