## 4. Mini Project Extension: Business Intelligence & Sales Analytics Engine
We implement executive-level analytics:
- Pareto top drivers vs. trailing inventory
- Category gross transaction value (GTV) and percentage contribution
- Cross-tabulated quarterly sales volume
- Automated Executive KPI audit

```python
# Category Performance Summary
total_rev = df['SalesAmount'].sum()
cat_perf = df.groupby('Category').agg(
    TotalSales=('SalesAmount', 'sum'),
    MeanSales=('SalesAmount', 'mean'),
    TotalUnits=('UnitsSold', 'sum'),
    AvgRating=('Rating', 'mean'),
    SKUCount=('ProductID', 'count')
).reset_index()
cat_perf['RevenueSharePct'] = (cat_perf['TotalSales'] / total_rev) * 100
cat_perf = cat_perf.sort_values(by='TotalSales', ascending=False)
display(cat_perf)

# Quarterly Cross-Tabulation Matrix
q_pivot = df.pivot_table(index='Quarter', columns='Category', values='SalesAmount', aggfunc='sum', fill_value=0)
display(q_pivot)
```

## 5. Visualizations & BI Dashboard
Comprehensive four-panel visual analytics cockpit.

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('E-Commerce Sales Analytics & Business Intelligence Dashboard\nVEDANT NIMKAR (26MML0045) | Week 1 Mini Project', fontsize=15, weight='bold', y=0.98)

# Subplot 1: Top 10 Products Horizontal Bar Chart
top_10 = sorted_df.head(10).sort_values(by='SalesAmount', ascending=True)
colors = plt.cm.viridis(np.linspace(0.3, 0.85, len(top_10)))
bars = axes[0, 0].barh(top_10['ProductName'], top_10['SalesAmount'] / 1000, color=colors, edgecolor='black', alpha=0.85)
axes[0, 0].set_title('Top 10 Products by Sales Revenue (in Thousands INR)', weight='bold')
axes[0, 0].set_xlabel("Sales Revenue (₹ '000)")
for bar in bars:
    w = bar.get_width()
    axes[0, 0].text(w + 1, bar.get_y() + bar.get_height()/2, f'₹{w:.0f}k', va='center', fontsize=9, weight='bold')

# Subplot 2: Category Revenue Share Donut Chart
cat_palette = sns.color_palette('Set2', len(cat_perf))
wedges, texts, autotexts = axes[0, 1].pie(
    cat_perf['TotalSales'], labels=cat_perf['Category'], autopct='%1.1f%%',
    startangle=140, colors=cat_palette, pctdistance=0.75,
    wedgeprops=dict(width=0.45, edgecolor='white', linewidth=2)
)
plt.setp(autotexts, size=9, weight='bold')
axes[0, 1].set_title('Revenue Contribution Share by Category', weight='bold')

# Subplot 3: Quarterly Sales Trends by Category
q_pivot.plot(kind='bar', stacked=True, ax=axes[1, 0], colormap='tab10', edgecolor='black', alpha=0.85)
axes[1, 0].set_title('Quarterly Revenue Accumulation by Category', weight='bold')
axes[1, 0].set_xlabel('Fiscal Quarter')
axes[1, 0].set_ylabel('Total Sales (₹)')
axes[1, 0].legend(title='Category', bbox_to_anchor=(1.02, 1), loc='upper left')
axes[1, 0].tick_params(axis='x', rotation=0)

# Subplot 4: Customer Rating vs Sales Amount (Bubble Chart sized by Units Sold)
scatter = axes[1, 1].scatter(
    df['Rating'], df['SalesAmount'] / 1000, s=df['UnitsSold']*3,
    c=df['SalesAmount'], cmap='plasma', alpha=0.75, edgecolors='black', linewidth=1.2
)
axes[1, 1].set_title('Product Rating vs. Sales Revenue (Bubble Size = Units Sold)', weight='bold')
axes[1, 1].set_xlabel('Customer Rating (1.0 to 5.0 Stars)')
axes[1, 1].set_ylabel("Sales Revenue (₹ '000)")
cbar = plt.colorbar(scatter, ax=axes[1, 1])
cbar.set_label('Sales Amount (₹)')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
