## Problem Statement
Create an interactive bar chart showing the sales figures of different products using Bokeh. Formulate a retail sales dataset with multiple product lines, quarterly revenue figures, units sold, and profit margins. Render an interactive Bokeh vertical bar chart using p.vbar(), apply categorical color palettes, add hover tooltips showing units sold and revenue in Lakhs, and style canvas axes, titles, and grid lines.

## Objectives
- To construct interactive categorical bar charts using Bokeh's p.vbar() glyph.
- To manage categorical coordinate systems and factor ranges in Bokeh.
- To apply color palettes to distinguish individual product categories.
- To integrate an interactive HoverTool displaying revenue, units sold, and profit margins.
- To calculate sales contribution percentages and ranking metrics using Pandas.
- To analyze retail product performance and identify top-performing revenue drivers.


```python
# --- Cell 1: Product Sales Dataset, Revenue Rankings, and Portfolio Share ---
# Load Necessary Packages
import pandas as pd
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.transform import factor_cmap

# Step 1 & 2: Retail sales data formulation
sales_records = [
    {"Product": "Smartphones", "Category": "Mobile Devices", "Units_Sold": 1450, "Revenue_Lakhs": 217.5, "Margin_Pct": 18.5},
    {"Product": "Laptops", "Category": "Computing", "Units_Sold": 680, "Revenue_Lakhs": 183.6, "Margin_Pct": 14.2},
    {"Product": "Smart TVs", "Category": "Home Entertainment", "Units_Sold": 420, "Revenue_Lakhs": 105.0, "Margin_Pct": 16.8},
    {"Product": "Tablets", "Category": "Mobile Devices", "Units_Sold": 530, "Revenue_Lakhs": 68.9, "Margin_Pct": 15.0},
    {"Product": "Gaming Consoles", "Category": "Gaming", "Units_Sold": 310, "Revenue_Lakhs": 62.0, "Margin_Pct": 12.5},
    {"Product": "Smartwatches", "Category": "Wearables", "Units_Sold": 720, "Revenue_Lakhs": 43.2, "Margin_Pct": 22.0},
    {"Product": "Headphones", "Category": "Audio", "Units_Sold": 1100, "Revenue_Lakhs": 33.0, "Margin_Pct": 28.5},
    {"Product": "Wireless Speakers", "Category": "Audio", "Units_Sold": 650, "Revenue_Lakhs": 19.5, "Margin_Pct": 25.0}
]

df_sales = pd.DataFrame(sales_records)
df_sales = df_sales.sort_values("Revenue_Lakhs", ascending=False).reset_index(drop=True)

total_rev = df_sales["Revenue_Lakhs"].sum()
df_sales["Revenue_Share_Pct"] = ((df_sales["Revenue_Lakhs"] / total_rev) * 100).round(1)

print("=== Retail Product Sales & Revenue Rankings ===")
print(df_sales.to_string(index=False))

print(f"\nTotal Portfolio Revenue: ₹{total_rev:.1f} Lakhs")
print(f"Top 2 Categories Contribution: {df_sales.loc[:1, 'Revenue_Share_Pct'].sum():.1f}% of total sales")
```

```python
# --- Cell 2: Interactive Bokeh Vertical Bar Chart with Category Color Mapping ---
# Step 4 to 8: Interactive Bokeh vertical bar chart
source = ColumnDataSource(df_sales)
product_names = df_sales["Product"].tolist()
categories = df_sales["Category"].unique().tolist()
palette = ["#2b6cb0", "#805ad5", "#dd6b20", "#319795", "#d69e2e", "#e53e3e"]

color_mapping = factor_cmap("Category", palette=palette[:len(categories)], factors=categories)

hover = HoverTool(tooltips=[
    ("Product", "@Product"),
    ("Category", "@Category"),
    ("Revenue", "₹@Revenue_Lakhs{0.0} Lakhs"),
    ("Units Sold", "@Units_Sold units"),
    ("Profit Margin", "@Margin_Pct{0.0}%"),
    ("Revenue Share", "@Revenue_Share_Pct{0.0}%")
])

p = figure(
    title="Product Revenue Distribution Across Retail Electronics Lines (Bokeh)",
    x_range=product_names,
    x_axis_label="Product Line",
    y_axis_label="Revenue (Lakhs ₹)",
    width=680,
    height=440,
    tools=[hover, "pan", "wheel_zoom", "reset"]
)

p.vbar(
    x="Product",
    top="Revenue_Lakhs",
    source=source,
    width=0.6,
    color=color_mapping,
    legend_field="Category",
    line_color="#1a202c",
    line_width=0.8
)

p.title.text_font_size = "13pt"
p.title.align = "center"
p.xaxis.major_label_orientation = 0.52
p.y_range.start = 0
p.y_range.end = 245
p.legend.title = "Category"
p.legend.location = "top_right"
p.legend.click_policy = "hide"
p.background_fill_color = "#fbfbfb"
p.grid.grid_line_alpha = 0.4

show(p)
```
