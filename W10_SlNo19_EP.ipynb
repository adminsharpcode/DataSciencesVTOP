## Problem Statement
Create a list of the first five prime numbers. Write a function that takes this list and returns a new list with each element squared. Then, create a set of unique characters from the string 'hello world' and write a function that takes two sets and returns their intersection. Create a tuple with three elements: an integer, a string, and a boolean, then access and print each element.

## Objectives
- To explore immutable, singly linked list data structures and linear recursive processing in Scala.
- To implement functional list transformations using the higher-order map combinator.
- To construct mathematical sets and execute set-theoretic operations such as intersection.
- To construct and unpack heterogeneous, type-safe tuples with positional member accessors.
- To contrast the structural and operational characteristics of Lists, Sets, and Tuples in Scala.


```scala
# --- Cell 1: Prime List Squaring and Set Intersection Operations ---
// Program 19: Collections in Scala: Lists, Sets, and Tuples
// Student: Vedant Nimkar | Reg. No: 26MML0045

object Experiment19_ListsSetsTuples {
  def main(args: Array[String]): Unit = {
    // 1. Prime List & Functional Squaring
    val primes: List[Int] = List(2, 3, 5, 7, 11)

    def squareList(lst: List[Int]): List[Int] = {
      lst.map(x => x * x)
    }

    println(s"Original primes: $primes")
    println(s"Squared primes:  ${squareList(primes)}")

    // 2. Set of Unique Characters & Set Intersection
    val charSet1: Set[Char] = "hello world".toSet
    val charSet2: Set[Char] = "world wide web".toSet

    def setIntersection[T](s1: Set[T], s2: Set[T]): Set[T] = {
      s1 & s2
    }

    println(s"Unique chars in 'hello world': $charSet1")
    println(s"Intersection with 'world wide web': ${setIntersection(charSet1, charSet2)}")
  }
}
```

## Conclusion
The Scala collections experiment demonstrated the structural diversity, functional purity, and type safety afforded by Lists, Sets, and Tuples.

- Scala Lists are persistent immutable linked lists optimized for prepend and linear traversal operations.
- The higher-order map method provides expressive element-wise transformations without mutating source data.
- Sets automatically guarantee uniqueness of elements and enable rapid mathematical set operations like intersection.
- Tuples facilitate lightweight grouping of heterogeneous values with strict compile-time type inference.
- Immutable collection semantics eliminate concurrency hazards in multi-threaded data pipelines.

Proficiency with Scala's collection hierarchy is essential for building robust, idiomatic functional programs and real-time distributed data pipelines.
