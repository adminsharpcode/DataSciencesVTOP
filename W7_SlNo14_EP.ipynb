## Problem Statement
Create an interactive line plot showing the daily temperatures over a week using Bokeh. Build a dataset recording Maximum Temperature, Minimum Temperature, and Average Temperature across seven days of a week. Construct an interactive Bokeh multi-line plot with distinct stroke colors, circle marker glyphs at data points, interactive hover tooltips reporting daily temperatures, and custom grid styling.

## Objectives
- To visualize time-series and sequential daily meteorological data using Bokeh line glyphs.
- To combine multiple visual glyphs (line and circle) on the same canvas for enhanced readability.
- To plot multiple temperature series (Max, Min, Avg) with custom palettes and line dash styles.
- To configure a multi-line HoverTool displaying day name and corresponding temperature readings.
- To customize grid lines, background fill, axis labels, and interactive legend placement.
- To analyze diurnal temperature ranges and weekly atmospheric temperature stability.


```python
# --- Cell 1: Weekly Daily Temperature Measurements and Diurnal Metrics ---
# Load Necessary Packages
import pandas as pd
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, HoverTool

# Step 1 & 2: Define weekly meteorological readings
weather_data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Date": ["17-08-2026", "18-08-2026", "19-08-2026", "20-08-2026", "21-08-2026", "22-08-2026", "23-08-2026"],
    "Max_Temp_C": [33.5, 34.2, 35.6, 32.8, 31.0, 32.5, 33.0],
    "Min_Temp_C": [24.0, 24.8, 25.2, 23.5, 22.2, 23.0, 23.8],
    "Avg_Temp_C": [28.8, 29.5, 30.4, 28.2, 26.6, 27.8, 28.4]
}

df_weather = pd.DataFrame(weather_data)
df_weather["Diurnal_Range_C"] = (df_weather["Max_Temp_C"] - df_weather["Min_Temp_C"]).round(1)

print("=== Weekly Daily Temperatures Dataset ===")
print(df_weather.to_string(index=False))

# Step 3: Weekly meteorological summary
print("\n=== Weekly Meteorological Summary (°C) ===")
print(f"Weekly Average Daytime High: {df_weather['Max_Temp_C'].mean():.2f}°C")
print(f"Weekly Average Nighttime Low: {df_weather['Min_Temp_C'].mean():.2f}°C")
print(f"Peak Recorded High: {df_weather['Max_Temp_C'].max():.1f}°C (on {df_weather.loc[df_weather['Max_Temp_C'].idxmax(), 'Day']})")
print(f"Lowest Recorded Low: {df_weather['Min_Temp_C'].min():.1f}°C (on {df_weather.loc[df_weather['Min_Temp_C'].idxmin(), 'Day']})")
print(f"Mean Diurnal Spread: {df_weather['Diurnal_Range_C'].mean():.2f}°C")
```

```python
# --- Cell 2: Interactive Bokeh Multi-Line Plot of Daily Temperatures with Markers ---
# Step 4 to 8: Multi-line interactive Bokeh temperature plot
source = ColumnDataSource(df_weather)
days_order = df_weather["Day"].tolist()

hover = HoverTool(tooltips=[
    ("Day", "@Day (@Date)"),
    ("Max Temp", "@Max_Temp_C{0.0}°C"),
    ("Min Temp", "@Min_Temp_C{0.0}°C"),
    ("Average", "@Avg_Temp_C{0.0}°C"),
    ("Diurnal Spread", "@Diurnal_Range_C{0.0}°C")
])

p = figure(
    title="Daily Temperatures Over a Week (Monday – Sunday)",
    x_range=days_order,
    x_axis_label="Day of the Week",
    y_axis_label="Temperature (°C)",
    width=680,
    height=440,
    tools=[hover, "pan", "wheel_zoom", "reset"]
)

# Render Maximum Temperature line & markers
p.line(x="Day", y="Max_Temp_C", source=source, line_width=2.5, color="#c53030", legend_label="Max Temperature")
p.circle(x="Day", y="Max_Temp_C", source=source, size=8, fill_color="#c53030", line_color="white", line_width=1)

# Render Average Temperature line & markers
p.line(x="Day", y="Avg_Temp_C", source=source, line_width=2, color="#2b6cb0", line_dash="dashed", legend_label="Avg Temperature")
p.square(x="Day", y="Avg_Temp_C", source=source, size=7, fill_color="#2b6cb0", line_color="white", line_width=1)

# Render Minimum Temperature line & markers
p.line(x="Day", y="Min_Temp_C", source=source, line_width=2.5, color="#2f855a", legend_label="Min Temperature")
p.circle(x="Day", y="Min_Temp_C", source=source, size=8, fill_color="#2f855a", line_color="white", line_width=1)

p.title.text_font_size = "13pt"
p.title.align = "center"
p.y_range.start = 18
p.y_range.end = 39
p.legend.location = "top_right"
p.legend.click_policy = "hide"
p.background_fill_color = "#fbfbfb"
p.grid.grid_line_alpha = 0.45

show(p)
```
