## Problem Statement
Create a DataFrame with columns Name, Age, and City containing data for five individuals. Perform the following operations: select only the Name and Age columns, filter rows where Age is greater than 25, add a new column Country with a default value, and sort the DataFrame by Age in descending order. Finally, calculate the average age of the individuals.

## Objectives
- To understand how to create a Pandas DataFrame from a Python dictionary.
- To learn column selection techniques using Pandas.
- To apply filtering conditions on DataFrame rows.
- To create and populate new columns in a DataFrame.
- To sort data based on column values.
- To perform basic statistical analysis using Pandas aggregate functions.


```python
# --- Cell 1: Original DataFrame ---
# Load Necessary Packages
import pandas as pd

# Step 1: Create dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma'],
    'Age': [23, 30, 28, 35, 22],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney']
}

# Step 2: Create DataFrame
df = pd.DataFrame(data)

print("Original DataFrame")
print(df)
```

```python
# --- Cell 2: Selected Name and Age Columns ---
# Step 3: Select Name and Age columns
name_age = df[['Name', 'Age']]
print("\nName and Age")
print(name_age)
```

```python
# --- Cell 3: Filtered Data (Age > 25) ---
# Step 4: Filter Age > 25
filtered_df = df[df['Age'] > 25]
print("\nAge > 25")
print(filtered_df)
```

```python
# --- Cell 4: DataFrame with Country Column ---
# Step 5: Add Country column
df['Country'] = 'India'
print("\nCountry Added")
print(df)
```

```python
# --- Cell 5: Sorted Data by Age (Descending Order) ---
# Step 6: Sort by Age in descending order
sorted_df = df.sort_values(by='Age', ascending=False)
print("\nSorted Data")
print(sorted_df)
```

```python
# --- Cell 6: Average Age of Individuals ---
# Step 7: Calculate Average Age
average_age = df['Age'].mean()
print("\nAverage Age =", average_age)
```
