## Problem Statement
Create a DataFrame with some missing values in the columns Name, Age, and City. Perform the following operations: fill the missing values in the Age column with the mean age, drop the rows where Name or City is missing, and convert the Age column to integer type. Finally, normalize the Age column using min-max scaling.

## Objectives
- To understand the importance of data cleaning and preprocessing using Pandas.
- To identify and count the missing (null) values present in a dataset.
- To replace missing numerical values using the mean of the column.
- To remove incomplete records in which key categorical fields are missing.
- To convert the data type of a column using type casting.
- To apply min-max scaling in order to normalize numerical data.


```python
# --- Cell 1: Original DataFrame with Missing Values ---
# Load Necessary Packages
import pandas as pd
import numpy as np

# Step 1: Create a dataset containing missing values
data = {
    'Name': ['Alice', 'Bob', None, 'David', 'Emma', 'Frank', None, 'Helen'],
    'Age': [23, np.nan, 28, 35, np.nan, 41, 30, 26],
    'City': ['New York', 'London', 'Paris', None, 'Sydney',
             'Berlin', 'Tokyo', None]
}

# Step 2: Create the DataFrame
df = pd.DataFrame(data)

print("Original DataFrame with Missing Values")
print(df)
```

```python
# --- Cell 2: Count of Missing Values in Each Column ---
# Step 3: Identify the missing values in each column
print("\nCount of Missing Values in Each Column")
print(df.isnull().sum())
print("\nTotal number of missing values =", df.isnull().sum().sum())
```

```python
# --- Cell 3: Missing Age Values Replaced by the Mean Age ---
# Step 4: Fill the missing Age values with the mean age
mean_age = df['Age'].mean()
print("\nMean Age used for imputation =", mean_age)

df['Age'] = df['Age'].fillna(mean_age)
print("\nDataFrame after filling missing Age with the mean")
print(df)
```

```python
# --- Cell 4: DataFrame after Dropping Rows with Missing Name or City ---
# Step 5: Drop the rows where Name or City is missing
df = df.dropna(subset=['Name', 'City'])
print("\nDataFrame after dropping rows with missing Name or City")
print(df)
print("\nShape after dropping incomplete rows =", df.shape)
```

```python
# --- Cell 5: Age Column Converted to Integer Type ---
# Step 6: Convert the Age column to integer type
df['Age'] = df['Age'].round().astype(int)
print("\nDataFrame after converting Age to integer type")
print(df)
print("\nData types of the columns")
print(df.dtypes)
```

```python
# --- Cell 6: Final Cleaned DataFrame with Min-Max Normalized Age ---
# Step 7: Normalize the Age column using Min-Max scaling
age_min = df['Age'].min()
age_max = df['Age'].max()
df['Age_Normalized'] = (df['Age'] - age_min) / (age_max - age_min)

print("\nMinimum Age =", age_min, " Maximum Age =", age_max)
print("\nFinal Cleaned and Normalized DataFrame")
print(df)
```

## Conclusion
