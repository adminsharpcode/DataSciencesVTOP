## Problem Statement
Create a case class Person with fields name, age, and city. Create a list of Person objects representing five different individuals. Write a function to filter out people older than 30. Next, write a function that groups people by their city. Then, write a function that transforms each Person object into a string in the format 'Name is Age years old and lives in City'. Finally, write a function to sort the list of people by age in ascending order.

## Objectives
- To master Domain-Driven Design and immutable data modeling using Scala Case Classes.
- To understand compiler-generated methods in case classes: apply, copy, equals, hashCode, and toString.
- To implement collection filtering using domain entity attributes (age > 30).
- To perform multi-way grouping using the groupBy combinator on case class fields.
- To construct formatted string descriptions using string interpolation and perform sorting with sortBy.


```scala
# --- Cell 1: Case Class Instantiation, Field-Based Filtering, and Multi-Way Grouping ---
// Program 22: Case Classes and Data Modeling in Scala
// Student: Vedant Nimkar | Reg. No: 26MML0045

case class Person(name: String, age: Int, city: String)

object Experiment22_CaseClasses {
  def main(args: Array[String]): Unit = {
    val people = List(
      Person("Vedant Nimkar", 24, "Mumbai"),
      Person("Ayush Sharma", 32, "Delhi"),
      Person("Rohan Yadav", 28, "Bengaluru"),
      Person("Priya Patel", 35, "Mumbai"),
      Person("Sneha Reddy", 22, "Hyderabad")
    )

    // 1. Filter out people older than 30
    val olderThan30 = people.filter(_.age > 30)
    println("1. People Older than 30:")
    olderThan30.foreach(p => println(s"   - ${p.name} (Age: ${p.age})"))

    // 2. Group people by city
    val groupedByCity = people.groupBy(_.city)
    println("\n2. Grouped by City:")
    groupedByCity.foreach {
      case (city, group) => println(s"   - $city: ${group.map(_.name).mkString(", ")}")
    }
  }
}
```

## Conclusion
The Scala Case Classes experiment validated the paradigm of Domain-Driven Modeling and algebraic data types in modern software engineering.

- Case classes eliminate boilerplate by automatically generating constructors, getters, toString, and equals.
- Immutability of case class fields guarantees thread safety and prevents unintentional state corruption.
- Higher-order combinators like filter and groupBy operate seamlessly on typed case class properties.
- String interpolation enables expressive, type-checked textual rendering of domain models.
- Case classes integrate seamlessly with pattern matching and serialization libraries across the Scala ecosystem.

Case classes are the architectural cornerstone of data pipelines in Apache Spark, defining structured schemas for strongly-typed Datasets.
