## Problem Statement
A manufacturing company operates 4 production lines across 5 shifts every day and the management wants to analyse production efficiency and identify underperforming production lines. Create a 4 x 5 NumPy array representing the production units manufactured by each production line during the different shifts and perform the following operations: calculate the mean production units across all shifts and production lines, calculate the standard deviation to measure production variability, slice the first 3 production lines and the last 2 shifts, reshape the array into a 5 x 4 matrix, and create another 5 x 4 machine efficiency matrix and perform element-wise multiplication.

## Objectives
- To represent factory production data of 4 production lines and 5 shifts as a NumPy matrix.
- To compute the mean production output as a benchmark for factory performance.
- To measure production variability across lines and shifts using standard deviation.
- To extract a specific block of production data using array slicing.
- To reshape the production matrix from 4 x 5 into 5 x 4 form.
- To apply a machine efficiency matrix through element-wise multiplication in order to obtain the effective production output.


```python
# --- Cell 1: Production Matrix of 4 Production Lines across 5 Shifts ---
# Load Necessary Packages
import numpy as np

# Fix the random seed for reproducible results
np.random.seed(42)

# Step 1: Create the 4 x 5 production matrix
# Rows    -> 4 production lines
# Columns -> 5 shifts
production = np.array([
    [120, 135, 128, 142, 150],   # Line 1
    [ 98, 110, 105, 115, 120],   # Line 2
    [160, 155, 170, 165, 158],   # Line 3
    [ 88,  92,  85,  95,  90]    # Line 4
])

print("Production Matrix (4 Lines x 5 Shifts)")
print(production)
print("\nShape =", production.shape)
print("Total production units =", production.sum())
```

```python
# --- Cell 2: Mean and Standard Deviation of Production Units ---
# Step 2: Calculate the mean production units across all lines and shifts
mean_production = production.mean()

# Step 3: Calculate the standard deviation to measure production variability
std_production = production.std()

print("\nMean production units (all lines and shifts) =", mean_production)
print("Standard deviation of production units       =", std_production)
```

```python
# --- Cell 3: Line-wise Average Output and Underperforming Line ---
# Step 4: Line-wise average output and identification of the weakest line
line_averages = production.mean(axis=1)
print("\nAverage output of each production line")
for i, avg in enumerate(line_averages, start=1):
    print("Line", i, "average =", avg)

weakest = line_averages.argmin()
print("\nUnderperforming production line = Line", weakest + 1,
      "with average", line_averages[weakest])
```

```python
# --- Cell 4: Sliced Production Data (First 3 Lines, Last 2 Shifts) ---
# Step 5: Slice the first 3 production lines and the last 2 shifts
sliced = production[:3, -2:]

print("\nSliced Matrix (first 3 lines, last 2 shifts)")
print(sliced)
print("\nShape of the sliced matrix =", sliced.shape)
```

```python
# --- Cell 5: Production Matrix Reshaped from 4 x 5 into 5 x 4 ---
# Step 6: Reshape the 4 x 5 production matrix into a 5 x 4 matrix
reshaped = production.reshape(5, 4)

print("\nReshaped Production Matrix (5 x 4)")
print(reshaped)
print("\nShape after reshaping =", reshaped.shape)
```

```python
# --- Cell 6: Machine Efficiency Matrix and Effective Production Output ---
# Step 7: Create a 5 x 4 machine efficiency matrix
efficiency = np.random.randint(1, 4, size=(5, 4))
print("\nMachine Efficiency Matrix (5 x 4)")
print(efficiency)

# Step 8: Element-wise multiplication
effective_output = reshaped * efficiency
print("\nEffective Production Output (element-wise multiplication)")
print(effective_output)
print("\nTotal effective output =", effective_output.sum())
```
