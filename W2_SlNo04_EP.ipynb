## Problem Statement
A transportation company maintains fleet records containing VehicleID, FuelConsumption, and Route, in which some records contain missing fuel consumption values and route information. Perform the following operations: create a DataFrame containing missing values, fill the missing FuelConsumption values using the mean FuelConsumption, remove the rows where VehicleID or Route is missing, convert FuelConsumption to integer type, and normalize FuelConsumption using min-max scaling.

## Objectives
- To create a fleet management DataFrame containing missing values.
- To detect and count the missing entries in the fleet records.
- To impute the missing FuelConsumption readings using the mean value.
- To eliminate the records in which VehicleID or Route information is unavailable.
- To convert the FuelConsumption column into integer type using type casting.
- To normalize FuelConsumption using min-max scaling for fleet efficiency comparison.


```python
# --- Cell 1: Original Fleet DataFrame with Missing Values ---
# Load Necessary Packages
import pandas as pd
import numpy as np

# Step 1: Create the fleet dataset containing missing values
data = {
    'VehicleID': ['V001', 'V002', None, 'V004', 'V005',
                  'V006', 'V007', None, 'V009', 'V010'],
    'FuelConsumption': [12.5, np.nan, 15.2, 9.8, np.nan,
                        14.1, 11.6, 13.3, np.nan, 10.4],
    'Route': ['Chennai-Vellore', 'Vellore-Bangalore', 'Chennai-Trichy', None,
              'Madurai-Chennai', 'Vellore-Chennai', None,
              'Coimbatore-Salem', 'Salem-Erode', 'Erode-Trichy']
}

# Step 2: Create the DataFrame
df = pd.DataFrame(data)

print("Original Fleet DataFrame with Missing Values")
print(df)
```

```python
# --- Cell 2: Count of Missing Values in the Fleet Records ---
# Step 3: Identify the missing values in each column
print("\nCount of Missing Values in Each Column")
print(df.isnull().sum())
print("\nTotal number of missing values =", df.isnull().sum().sum())
```

```python
# --- Cell 3: Missing FuelConsumption Replaced by the Mean Value ---
# Step 4: Fill the missing FuelConsumption with the mean value
mean_fuel = df['FuelConsumption'].mean()
print("\nMean FuelConsumption used for imputation =", mean_fuel)

df['FuelConsumption'] = df['FuelConsumption'].fillna(mean_fuel)
print("\nFleet DataFrame after filling missing FuelConsumption")
print(df)
```

```python
# --- Cell 4: Fleet Records after Removing Missing VehicleID or Route ---
# Step 5: Remove the rows where VehicleID or Route is missing
df = df.dropna(subset=['VehicleID', 'Route'])
print("\nFleet DataFrame after removing incomplete records")
print(df)
print("\nShape after removing incomplete records =", df.shape)
```

```python
# --- Cell 5: FuelConsumption Converted to Integer Type ---
# Step 6: Convert FuelConsumption to integer type
df['FuelConsumption'] = df['FuelConsumption'].round().astype(int)
print("\nFleet DataFrame after converting FuelConsumption to integer")
print(df)
print("\nData types of the columns")
print(df.dtypes)
```

```python
# --- Cell 6: Final Normalized Fleet Data and Most Efficient Vehicle ---
# Step 7: Normalize FuelConsumption using Min-Max scaling
fuel_min = df['FuelConsumption'].min()
fuel_max = df['FuelConsumption'].max()
df['Fuel_Normalized'] = (df['FuelConsumption'] - fuel_min) / (fuel_max - fuel_min)

print("\nMinimum =", fuel_min, " Maximum =", fuel_max)
print("\nFinal Cleaned and Normalized Fleet DataFrame")
print(df)

# Most fuel-efficient vehicle (lowest consumption)
best = df.loc[df['FuelConsumption'].idxmin()]
print("\nMost fuel-efficient vehicle:")
print(best)
```
