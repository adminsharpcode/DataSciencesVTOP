## Problem Statement
Create a HashMap with keys representing student names and values representing their grades, then insert multiple entries including a duplicate key with a different grade. Write a function to merge two HashMap’s, resolving conflicts by taking the higher grade. Finally, write a function to handle collisions by chaining, using lists to store multiple values for a single key.

## Objectives
- To comprehend hash table architectures, bucket hashing, and key conflict resolution strategies in Scala.
- To construct and populate mutable and immutable Map/HashMap data structures.
- To implement custom collision resolution logic when merging associative maps with duplicate keys.
- To build a multimap via separate chaining using List collections to preserve all conflicting values.
- To analyze algorithmic time complexity for hash lookups, insertions, and conflict management.


```scala
# --- Cell 1: HashMap Merge with Maximum Value Conflict Resolution ---
// Program 20: HashMaps and Conflict Resolution in Scala
// Student: Vedant Nimkar | Reg. No: 26MML0045

import scala.collection.mutable

object Experiment20_HashMaps {
  def main(args: Array[String]): Unit = {
    // 1. Merge two HashMaps resolving conflicts by taking higher grade
    val map1 = Map("Vedant" -> 88, "Rohan" -> 92, "Ayush" -> 78, "Priya" -> 84)
    val map2 = Map("Vedant" -> 95, "Rohan" -> 85, "Kalyan" -> 90, "Priya" -> 89)

    def mergeResolveMax(m1: Map[String, Int], m2: Map[String, Int]): Map[String, Int] = {
      (m1.keySet ++ m2.keySet).map { k =>
        val grade1 = m1.getOrElse(k, 0)
        val grade2 = m2.getOrElse(k, 0)
        k -> Math.max(grade1, grade2)
      }.toMap
    }

    val merged = mergeResolveMax(map1, map2)
    println("=== Student Grades HashMap Merge with Conflict Resolution ===")
    println(s"HashMap 1 (Original Grades): $map1")
    println(s"HashMap 2 (Updated / Re-eval): $map2")
    println("Merged HashMap (Higher Grade Selected):")
    merged.toSeq.sortBy(_._1).foreach { case (k, v) => println(s"  - $k -> $v") }
  }
}
```

```scala
# --- Cell 2: Collision Handling via List-Based Separate Chaining ---
    // 2. Collision Chaining via Multimap (List of values per key)
    val multiGradeMap = mutable.Map[String, List[Int]]()

    def insertChained(name: String, grade: Int): Unit = {
      val existing = multiGradeMap.getOrElse(name, Nil)
      multiGradeMap(name) = existing :+ grade
    }

    println("\n=== Chained Collision Resolution Multimap (List Chaining) ===")
    insertChained("Vedant", 88)
    insertChained("Vedant", 95) // Duplicate key appended to chain
    insertChained("Rohan", 92)
    insertChained("Rohan", 85)  // Duplicate key appended to chain
    insertChained("Ayush", 78)

    println("Final Chained Multimap Contents:")
    multiGradeMap.toSeq.sortBy(_._1).foreach {
      case (k, v) => println(s"  - $k : ${v.mkString("[", ", ", "]")}")
    }
```

## Conclusion
The Scala HashMap conflict resolution experiment validated the practical design patterns for managing key collisions and merging associative data structures.

- HashMaps provide near-constant O(1) expected time complexity for key insertion and retrieval operations.
- Set union over key sets enables comprehensive reconciliation across independent key-value datasets.
- Custom conflict resolution closures allow application-specific arbitration (such as selecting maximum scores).
- Separate chaining using List buckets prevents data loss when multiple transactions occur under the same key.
- Scala provides both immutable and mutable map implementations tailored to different architectural needs.

Understanding associative data structures and collision resolution strategies is vital for building robust indexing engines, caching tiers, and distributed key-value stores.
