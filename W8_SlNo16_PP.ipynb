## Problem Statement
Perform Principal Component Analysis (PCA) on the Iris dataset and visualize the first two principal components using an interactive 2D scatter plot in Plotly. Standardize the four botanical features using StandardScaler, fit a PCA model with 2 components, report the explained variance ratio for each component, assemble the projection into a DataFrame with true species labels, construct an interactive Plotly 2D scatter plot with hover details and confidence markers, and interpret the cluster separation.

## Objectives
- To understand dimensionality reduction principles using Principal Component Analysis (PCA).
- To preprocess multivariate numerical data using z-score feature scaling (StandardScaler).
- To compute eigenvalues and eigenvectors to project high-dimensional data onto orthogonal axes.
- To quantify information retention via explained variance ratios of PC1 and PC2.
- To visualize projected observations on an interactive 2D coordinate plane using Plotly.
- To evaluate the separability of botanical species in the reduced latent subspace.


```python
# --- Cell 1: Standardization, PCA Fit-Transform, Variance Explained, and Component Loadings ---
# Load Necessary Packages
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import plotly.express as px

# Step 1 & 2: Load and inspect Iris features
iris = load_iris()
feature_names = ["sepal_length", "sepal_width", "petal_length", "petal_width"]
species_map = {0: "setosa", 1: "versicolor", 2: "virginica"}

# Step 3: Standardize the feature space
scaler = StandardScaler()
X_scaled = scaler.fit_transform(iris.data)

# Step 4: Fit 2-Component PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Step 5 & 6: Create DataFrame with PCA coordinates
df_pca = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
df_pca["Species"] = [species_map[val] for val in iris.target]
df_pca["Sample_Index"] = range(1, len(df_pca) + 1)

print("=== PCA Projection DataFrame (First 5 Rows) ===")
print(df_pca.head(5).to_string(index=False))

# Explained variance ratios
var_ratio = pca.explained_variance_ratio_
print(f"\n=== Explained Variance Ratios ===")
print(f"PC1 Variance Explained: {var_ratio[0]*100:.2f}%")
print(f"PC2 Variance Explained: {var_ratio[1]*100:.2f}%")
print(f"Total Retained Variance (PC1 + PC2): {sum(var_ratio)*100:.2f}%")

# Feature Loadings
loadings = pd.DataFrame(
    pca.components_.T,
    columns=["PC1_Loading", "PC2_Loading"],
    index=feature_names
).round(4)
print("\n=== Principal Component Feature Loadings ===")
print(loadings)
```

```python
# --- Cell 2: Interactive Plotly 2D PCA Scatter Plot of Botanical Clusters ---
# Step 7 & 8: Interactive Plotly 2D PCA scatter plot
color_map = {
    "setosa": "#1f77b4",
    "versicolor": "#ff7f0e",
    "virginica": "#2ca02c"
}

fig = px.scatter(
    df_pca,
    x="PC1",
    y="PC2",
    color="Species",
    color_discrete_map=color_map,
    title="<b>PCA 2D Projection of Iris Dataset (95.8% Variance Retained)</b>",
    labels={
        "PC1": f"Principal Component 1 ({var_ratio[0]*100:.1f}% Variance)",
        "PC2": f"Principal Component 2 ({var_ratio[1]*100:.1f}% Variance)"
    },
    hover_name="Species",
    hover_data={"Sample_Index": True, "PC1": ":.2f", "PC2": ":.2f"}
)

fig.update_traces(
    marker=dict(size=10, opacity=0.85, line=dict(width=1, color="#1a202c"))
)

fig.update_layout(
    title_font_size=17,
    title_x=0.5,
    plot_bgcolor="#f8fafc",
    paper_bgcolor="white",
    xaxis_gridcolor="#e2e8f0",
    yaxis_gridcolor="#e2e8f0",
    margin=dict(l=55, r=40, t=60, b=60),
    legend=dict(title="Species", orientation="v", yanchor="top", y=0.98, xanchor="left", x=0.02)
)

fig.show()
```

## Conclusion
The PCA dimensionality reduction experiment confirmed the efficacy of unsupervised feature projection for multivariate cluster visualization.

- Z-score standardization successfully equalized variable scales across botanical features.
- Two principal components captured over 95.8% of the total dataset variance.
- Petal length and width contributed most significantly to the primary discriminative axis (PC1).
- Setosa achieved flawless separation in the 2D subspace without supervision.
- Plotly interactive scatter visualization provided intuitive exploration of cluster boundaries.

Dimensionality reduction via PCA drastically simplifies high-dimensional data analysis while preserving the fundamental geometric structure and cluster relationships.
