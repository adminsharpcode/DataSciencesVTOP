## Problem Statement
Perform K-means clustering on the Iris dataset and visualize the clusters using an interactive 3D scatter plot in Plotly. Standardize the Iris features, fit a KMeans clustering model with k=3, evaluate cluster quality using silhouette score and inertia, append predicted cluster labels to the dataset, render an interactive Plotly 3D scatter plot across three botanical dimensions (Sepal Length, Petal Length, Petal Width) color-coded by cluster assignment, add centroid markers, and compare unsupervised clusters against ground-truth species.

## Objectives
- To understand partition-based unsupervised learning using the K-Means clustering algorithm.
- To select and configure hyperparameter k=3 based on domain knowledge of botanical species.
- To fit K-Means on standardized data and identify optimal cluster centroids in feature space.
- To evaluate clustering cohesion and separation using inertia and the Silhouette Coefficient.
- To build an interactive 3D scatter plot using plotly.express.scatter_3d with orbital camera controls.
- To compare unsupervised cluster assignments against ground-truth botanical species labels.


```python
# --- Cell 1: K-Means Evaluation Metrics, Cross-Tabulation Matrix, and Centroids ---
# Load Necessary Packages
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import plotly.express as px

# Step 1 & 2: Load data and scale features
iris = load_iris()
feature_cols = ["sepal_length", "sepal_width", "petal_length", "petal_width"]
df_cluster = pd.DataFrame(iris.data, columns=feature_cols)
species_map = {0: "setosa", 1: "versicolor", 2: "virginica"}
df_cluster["True_Species"] = [species_map[i] for i in iris.target]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(iris.data)

# Step 3 & 4: Fit K-Means clustering with k=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled)
df_cluster["Cluster"] = [f"Cluster {c}" for c in cluster_labels]

# Step 5: Compute validation metrics
inertia = kmeans.inertia_
sil_score = silhouette_score(X_scaled, cluster_labels)

print("=== K-Means Clustering Model Evaluation ===")
print(f"Number of Clusters (k): 3")
print(f"Model Inertia (WCSS): {inertia:.2f}")
print(f"Overall Silhouette Score: {sil_score:.4f}")

# Step 6: Cross-tabulation comparing clusters with ground truth
print("\n=== Cluster Alignment with True Species (Confusion Matrix) ===")
ctab = pd.crosstab(df_cluster["Cluster"], df_cluster["True_Species"])
print(ctab)

# Unscale cluster centroids for physical interpretation
centroids_orig = scaler.inverse_transform(kmeans.cluster_centers_)
df_centroids = pd.DataFrame(centroids_orig, columns=feature_cols)
df_centroids.index = [f"Cluster {i}" for i in range(3)]
print("\n=== Cluster Centroids in Original Feature Units (cm) ===")
print(df_centroids.round(2))
```

```python
# --- Cell 2: Interactive Plotly 3D Scatter Plot of K-Means Partitioned Clusters ---
# Step 7 & 8: Interactive Plotly 3D scatter plot of clusters
color_discrete_map = {
    "Cluster 0": "#3182ce",
    "Cluster 1": "#d69e2e",
    "Cluster 2": "#e53e3e"
}

fig = px.scatter_3d(
    df_cluster,
    x="sepal_length",
    y="petal_length",
    z="petal_width",
    color="Cluster",
    color_discrete_map=color_discrete_map,
    title="<b>K-Means 3D Clustering Visualization on Iris Dataset (k=3)</b>",
    labels={
        "sepal_length": "Sepal Length (cm)",
        "petal_length": "Petal Length (cm)",
        "petal_width": "Petal Width (cm)"
    },
    hover_name="True_Species",
    hover_data={"Cluster": True, "sepal_length": ":.1f", "petal_length": ":.1f", "petal_width": ":.1f"}
)

fig.update_traces(
    marker=dict(size=5, opacity=0.88, line=dict(width=0.5, color="#1a202c"))
)

fig.update_layout(
    title_font_size=16,
    title_x=0.5,
    margin=dict(l=20, r=20, t=50, b=20),
    scene=dict(
        xaxis_title="Sepal Length (cm)",
        yaxis_title="Petal Length (cm)",
        zaxis_title="Petal Width (cm)",
        camera=dict(eye=dict(x=1.5, y=-1.5, z=1.2))
    ),
    legend=dict(title="Assigned Cluster", orientation="h", yanchor="bottom", y=0.02, xanchor="center", x=0.5)
)

fig.show()
```

## Conclusion
The K-Means clustering and 3D visualization experiment demonstrated the power of unsupervised partitioning algorithms coupled with multidimensional spatial rendering.

- Features were scaled with StandardScaler to prevent sepal length from dominating euclidean distances.
- K-Means with k=3 converged with an inertia of 139.82 and a silhouette score of 0.4599.
- Cluster 0 achieved 100% purity, perfectly identifying the Setosa botanical species.
- Centroid coordinates confirmed clear morphometric separation across petal dimensions.
- Plotly 3D scatter visualization provided orbital inspection of cluster boundaries in feature space.

Unsupervised clustering combined with interactive 3D visual exploration offers an intuitive and rigorous approach for discovering natural groupings in complex scientific datasets.
